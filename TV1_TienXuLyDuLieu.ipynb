```python
import pandas as pd

# Duong dan den du lieu MovieLens
duong_dan = "/content/drive/MyDrive/BTL_PTDL_DeTai10/0.1_Du_lieu_goc/ml-latest-small/"

# Doc hai file CSV
ratings = pd.read_csv(duong_dan + "ratings.csv")
movies = pd.read_csv(duong_dan + "movies.csv")

# Kiem tra kich thuoc du lieu
print("SO DONG VA SO COT")
print("ratings.csv:", ratings.shape)
print("movies.csv:", movies.shape)

# Xem 5 dong dau cua bang ratings
print("\n5 DONG DAU RATINGS:")
display(ratings.head())

# Xem 5 dong dau cua bang movies
print("\n5 DONG DAU MOVIES:")
display(movies.head())
```

```python
import pandas as pd

print("===== 1. KICH THUOC DU LIEU =====")
print("Ratings:", ratings.shape)
print("Movies:", movies.shape)

print("\n===== 2. DU LIEU BI THIEU =====")
print("Ratings:")
print(ratings.isnull().sum())

print("\nMovies:")
print(movies.isnull().sum())

print("\n===== 3. DU LIEU TRUNG LAP =====")
print("Ratings - so dong trung hoan toan:",
      ratings.duplicated().sum())

print("Movies - so dong trung hoan toan:",
      movies.duplicated().sum())

print("Movies - movieId trung:",
      movies["movieId"].duplicated().sum())

print("Ratings - cap userId, movieId trung:",
      ratings.duplicated(subset=["userId", "movieId"]).sum())

print("\n===== 4. KIEU DU LIEU =====")
print("Ratings:")
print(ratings.dtypes)

print("\nMovies:")
print(movies.dtypes)

print("\n===== 5. KIEM TRA DIEM RATING =====")
print("Diem nho nhat:", ratings["rating"].min())
print("Diem lon nhat:", ratings["rating"].max())

diem_khong_hop_le = ratings[
    ~ratings["rating"].isin(
        [0.5, 1.0, 1.5, 2.0, 2.5,
         3.0, 3.5, 4.0, 4.5, 5.0]
    )
]

print("So rating khong hop le:", len(diem_khong_hop_le))

print("\n===== 6. KIEM TRA MA PHIM =====")
ma_phim_khong_khop = ~ratings["movieId"].isin(movies["movieId"])
print("So danh gia khong tim thay phim:",
      ma_phim_khong_khop.sum())
```

```python
# BUOC 5.6: TAO BAN DU LIEU SACH

ratings_clean = ratings.copy()
movies_clean = movies.copy()

# Chuyen timestamp sang thoi gian de de phan tich sau nay
ratings_clean["rating_date"] = pd.to_datetime(
    ratings_clean["timestamp"],
    unit="s",
    utc=True
)

print("DA TAO BAN DU LIEU SACH")

print("\nKich thuoc ratings_clean:")
print(ratings_clean.shape)

print("\nKich thuoc movies_clean:")
print(movies_clean.shape)

print("\n5 dong dau ratings_clean:")
display(ratings_clean.head())

print("\n5 dong dau movies_clean:")
display(movies_clean.head())
```

```python
# BUOC 5.7: GHEP HAI BANG DU LIEU

# Ghep bang ratings va movies theo movieId
movie_data = pd.merge(
    ratings_clean,
    movies_clean,
    on="movieId",
    how="left",
    validate="many_to_one"
)

print("===== KET QUA GHEP DU LIEU =====")

print("So dong ratings truoc khi ghep:",
      len(ratings_clean))

print("So dong sau khi ghep:",
      len(movie_data))

print("Kich thuoc bang sau khi ghep:",
      movie_data.shape)

print("\nSo dong khong co ten phim:",
      movie_data["title"].isnull().sum())

print("\n5 DONG DAU SAU KHI GHEP:")
display(movie_data.head())

# Kiem tra so dong co duoc giu nguyen khong
assert len(movie_data) == len(ratings_clean), \
    "LOI: So dong sau khi ghep bi thay doi!"

assert movie_data["title"].notna().all(), \
    "LOI: Co danh gia khong tim thay ten phim!"

print("\nKIEM TRA THANH CONG: Khong mat du lieu!")
```

```python
# BUOC 5.8: THONG KE DANH GIA THEO PHIM

movie_summary = (
    movie_data
    .groupby(["movieId", "title", "genres"], as_index=False)
    .agg(
        rating_count=("rating", "count"),
        average_rating=("rating", "mean")
    )
)

# Lam tron diem trung binh den 2 chu so
movie_summary["average_rating"] = (
    movie_summary["average_rating"].round(2)
)

# Sap xep theo so luot danh gia giam dan
movie_summary = movie_summary.sort_values(
    by=["rating_count", "average_rating"],
    ascending=[False, False]
).reset_index(drop=True)

print("===== BUOC 5.8: KET QUA THONG KE =====")

print("So phim co it nhat 1 luot danh gia:",
      len(movie_summary))

print("Tong so luot danh gia:",
      movie_summary["rating_count"].sum())

print("\n10 PHIM DUOC DANH GIA NHIEU NHAT:")
display(movie_summary.head(10))

print("\nTHONG KE MO TA:")
display(movie_summary[
    ["rating_count", "average_rating"]
].describe())

# Kiem tra tong luot danh gia
assert movie_summary["rating_count"].sum() == len(movie_data)

print("\nBUOC 5.8 THANH CONG!")
```

```python
# BUOC 5.9: XUAT DU LIEU SACH

import os

output_folder = (
    "/content/drive/MyDrive/"
    "BTL_PTDL_DeTai10/0.2_Du_lieu_sach"
)

os.makedirs(output_folder, exist_ok=True)

# Luu cac bang du lieu
ratings_clean.to_csv(
    os.path.join(output_folder, "ratings_clean.csv"),
    index=False,
    encoding="utf-8-sig"
)

movies_clean.to_csv(
    os.path.join(output_folder, "movies_clean.csv"),
    index=False,
    encoding="utf-8-sig"
)

movie_data.to_csv(
    os.path.join(output_folder, "movie_data.csv"),
    index=False,
    encoding="utf-8-sig"
)

movie_summary.to_csv(
    os.path.join(output_folder, "movie_summary.csv"),
    index=False,
    encoding="utf-8-sig"
)

print("===== BUOC 5.9: XUAT FILE =====")

for file in [
    "ratings_clean.csv",
    "movies_clean.csv",
    "movie_data.csv",
    "movie_summary.csv"
]:
    path = os.path.join(output_folder, file)

    print(
        file,
        "- Ton tai:", os.path.isfile(path),
        "- Kich thuoc:",
        round(os.path.getsize(path) / 1024, 2),
        "KB"
    )

print("\nBUOC 5.9 HOAN THANH!")
```
